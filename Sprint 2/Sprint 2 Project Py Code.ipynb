```python
# 1. Import the required libraries
import numpy as np
from sklearn.ensemble import RandomForestClassifier

# 2. Input Features (Weather parameters)
# Format: [Temperature(°C), Humidity(%), Wind Speed(km/h), Rainfall(mm)]
X = [
    [38, 40, 15, 0],   # Summer
    [35, 45, 12, 2],   # Summer
    [27, 88, 25, 45],  # Monsoon / Rainy
    [25, 92, 30, 60],  # Monsoon / Rainy
    [15, 60, 10, 0],   # Winter
    [12, 55, 8, 0]     # Winter
]

# 3. Output Targets (Climate labels)
# 0 = Summer, 1 = Monsoon, 2 = Winter
y = [0, 0, 1, 1, 2, 2]

# 4. Dictionary to map numeric outputs to readable text
climate_names = {
    0: "Summer Season! ☀️",
    1: "Monsoon / Rainy Season! 🌧️",
    2: "Winter Season! ❄️"
}

# 5. Initialize and train the Random Forest Classifier
# n_estimators=100 creates an ensemble of 100 decision trees for high accuracy
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

print("--- 🤖 ML Model Training Completed! --- \n")

# 6. Test with completely new, unseen weather data
# Test Case 1: High temp, low humidity (Expected: Summer)
# Test Case 2: Cool temp, high humidity, heavy rain (Expected: Monsoon)
# Test Case 3: Very low temp, low rainfall (Expected: Winter)
new_weather_data = [
    [36, 42, 14, 0],   # Summer test
    [26, 90, 28, 50],  # Monsoon test
    [14, 52, 9, 1]     # Winter test
]

predictions = model.predict(new_weather_data)

# 7. Print the prediction results
print("--- 🌤️ Climate Prediction Results ---")
for i, data in enumerate(new_weather_data):
    print(f"Test Case {i+1}:")
    print(f"  -> Temperature: {data[0]}°C")
    print(f"  -> Humidity: {data[1]}%")
    print(f"  -> Wind Speed: {data[2]} km/h")
    print(f"  -> Rainfall: {data[3]} mm")
    
    predicted_class = predictions[i]
    print(f"🔮 Predicted Climate: {climate_names[predicted_class]}")
    print("-" * 40)
```
